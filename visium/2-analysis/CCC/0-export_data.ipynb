```r
library(Seurat)
library(SpatialExperiment)
library(HDF5Array)
```

```r
out_dir <- "~/SP-COVID/visium/2-analysis/CCC/out"
```

# Lung
## Visium

```r
spe <- loadHDF5SummarizedExperiment(dir = "~/SP-COVID/visium/0-processing/out/3-annot_lung_spe")

rownames(spatialCoords(spe)) <- paste(colData(spe)$cond_tissue, rownames(spatialCoords(spe)), sep = "_")
colnames(spe) <- paste(colData(spe)$cond_tissue, colnames(spe), sep = "_")

write.csv(as.data.frame(colData(spe)), glue::glue("{out_dir}/lung_data/visium/colData.csv"))
write.csv(as.data.frame(rowData(spe)), glue::glue("{out_dir}/lung_data/visium/rowData.csv"))

write.csv(as.data.frame(reducedDim(spe, "UMAP")), glue::glue("{out_dir}/lung_data/visium/UMAP.csv"))
write.csv(as.data.frame(reducedDim(spe, "PCA")), glue::glue("{out_dir}/lung_data/visium/PCA.csv"))
write.csv(as.data.frame(reducedDim(spe, "HARMONY")), glue::glue("{out_dir}/lung_data/visium/HARMONY.csv"))

write.csv(as.data.frame(spatialCoords(spe)), glue::glue("{out_dir}/lung_data/visium/spatialCoords.csv"))
```

## snRNA-seq

```r
se <- readRDS("~/SP-COVID/snRNAseq/0-annotation/out/se_annot_filtered.rds")
sce <- as.SingleCellExperiment(se)

write.csv(as.data.frame(colData(sce)), glue::glue("{out_dir}/lung_data/singlenuc/colData.csv"))

write.csv(as.data.frame(reducedDim(sce, "UMAP")), glue::glue("{out_dir}/lung_data/singlenuc/UMAP.csv"))
write.csv(as.data.frame(reducedDim(sce, "PCA")), glue::glue("{out_dir}/lung_data/singlenuc/PCA.csv"))
write.csv(as.data.frame(reducedDim(sce, "HARMONY")), glue::glue("{out_dir}/lung_data/singlenuc/HARMONY.csv"))
```

## Deconvolution matrix

```r
deconv <- readRDS("~/SP-COVID/visium/1-deconvolution/out/mat_filtered_lung.rds")

write.csv(as.data.frame(deconv), glue::glue("{out_dir}/lung_data/deconv_mat.csv"))
```
